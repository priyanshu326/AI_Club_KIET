## 4. Iteration: `for`, `while` and `range()`
Use a `for` loop when you want to process each item in a collection.

```python
items = [10, 20, 30, 40, 50]

for i in items:
    print(i)
```

### B. `while` loop — repeat while a condition is true

```python
count = 1
while count <= 5:
    print(count)
    count += 1
```

A `while` loop needs a starting value, a condition, and an update. Without a suitable update, the loop may never stop.

### C. Same task using both loops

```python
# for version
for number in range(1, 6):
    print(number)

# while version
number = 1
while number <= 5:
    print(number)
    number += 1
```

### D. `for` loop with `range()`

```python
# range(stop)
for number in range(5):
    print(number)
```

```python
# range(start, stop, step)
for number in range(1, 11, 2):
    print(number)
```

### E. Iterate over a list: values and indexes

```python
li = [10, 20, 30, 40, 50]

# Loop over values
for item in li:
    print(item)
```

```python
# Loop over indexes
for index in range(len(li)):
    print(index, li[index])
```

```python
# Use a while loop
index = 0
while index < len(li):
    print(li[index])
    index += 1
```

### B. Union, intersection and difference

```python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

print(A | B)  # union
print(A & B)  # intersection
print(A - B)  # values in A but not B

print(A.union(B))
print(A.intersection(B))
print(A.difference(B))
```
