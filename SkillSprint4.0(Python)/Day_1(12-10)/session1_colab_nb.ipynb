```python
#Question: Write a program that generates a random number between 1 and 10.
# Ask the user to guess it. Print "Correct guess!" if the guess matches;
# otherwise, print "Wrong guess!".

import random

number = random.randint(1, 10)
guess = int(input("Guess a number between 1 and 10: "))

if guess == number:
    print("Correct guess!")
else:
    print("Wrong guess!")

print("The number was:", number)

```

```python
#Generate a random integer between -10 and 10. Print whether the number is positive, negative, or zero.

import random

number = random.randint(-10, 10)

print("Number:", number)

if number > 0:
    print("Positive")
elif number < 0:
    print("Negative")
else:
    print("Zero")
```

```python
#Challenge: Random Number Range
#Question: Generate a random integer between 1 and 100. Print:
#- "Small" if it is less than 30.
#- "Medium" if it is between 30 and 70, inclusive.
#- "Large" if it is greater than 70.

import random

number = random.randint(1, 100)

print("Number:", number)

if number < 30:
    print("Small")
elif number <= 70:
    print("Medium")
else:
    print("Large")
```
