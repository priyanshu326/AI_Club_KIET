```python
#Types of Functions


 # 1)No parameters, no return value
def greet():
    print("Good morning!")
greet()


# 2) Parameters, no return value
def greet(name):
    print("Hello", name)

greet("Madhav")

# 3) No parameters, returns a value
def get_pi():
    return 3.14159

value = get_pi()
print(value)

#4) Parameters and a return value
def multiply(a, b):
    return a * b

result = multiply(4, 5)
print(result)
```

```python
class Car:
    def __init__(self, color, fuel):
        self.color = color
        self.speed = 0
        self.fuel = fuel

    def start(self):
        print("Car started")

    def accelerate(self):
        self.speed += 10

    def brake(self):
        self.speed = 0


car1 = Car("Red", 50)
car2 = Car("Blue", 70)

car1.accelerate()

print(car1.speed)  # 10
print(car2.speed)  # 0
```

```python
class Student:
    def __init__(self, name, branch):
        self.name = name
        self.branch = branch

    def introduce(self):
        print("I am", self.name)
        print("My branch is", self.branch)


s1 = Student("Madhav", "AIML")
s2 = Student("Rahul", "CSE")

s1.introduce()
s2.introduce()
```
